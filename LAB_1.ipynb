```python
import numpy as np

def add_vectors(vectora, vectorb):
    return np.add(vectora, vectorb)

def subtract_vectors(vectora, vectorb):
    return np.subtract(vectora, vectorb)

def multiply_vectors(vectora, vectorb):
    return np.multiply(vectora, vectorb)

# Example
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

print("Vector_a=",vector_a)
print("Vector_b=",vector_b)
result_addition = add_vectors(vector_a, vector_b)
result_subtraction = subtract_vectors(vector_a, vector_b)
result_multiplication = multiply_vectors(vector_a, vector_b)

print("Vector Addition:", result_addition)
print("Vector Subtraction:", result_subtraction)
print("Vector Multiplication:", result_multiplication)
```

```python
wv=[1,2,3,4,5,6,7,8,9]
vx=[1,9,8,4,5,6,7,3,2]
xv=np.random.rand(10)
```

```python
def v_v(wv, xv):
  total=0
  for w, x in zip(wv,vx):
    total += w*x
    return total
```

```python
w=xv
x=xv
z=v_v(w,x)
print(np.sum(z))
```
